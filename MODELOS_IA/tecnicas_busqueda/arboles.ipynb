```python
#recorrido en profundidad

def recorridoprofundidad (nodo_actual:Node):
    print(nodo_actual.name)
    for c in nodo_actual.children:
        recorridoprofundidad(c)

recorridoprofundidad(udo)
```

```python
def recorridoporanchura(nodoprimero:Node):
    porrecorrer=[nodoprimero]
    while porrecorrer:
        nodoactual=porrecorrer.pop(0)
        print(nodoactual.children)
        for hijo in nodoactual.children:
            porrecorrer.append(hijo)

recorridoporanchura(udo)
```
